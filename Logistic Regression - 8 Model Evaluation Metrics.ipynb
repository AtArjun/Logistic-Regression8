## 4. Data Perparation

### 4.1 Data Cleaning

```python
claimants_data.dropna(inplace=True)
claimants_data
```

### 4.2 Separation your Inputs and Outputs

```python
X = claimants_data[["CLMSEX","CLMINSUR","SEATBELT","CLMAGE","LOSS"]]
X
```

```python
X = claimants_data.drop(["CASENUM","ATTORNEY"], axis=1)
X
```

```python
y = claimants_data["ATTORNEY"]
y
```

## 5. Model Building

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=34)
```

```python
X_train
```

```python
X_test
```

## 6. Model Training

```python
logistic_model = LogisticRegression()
```

```python
logistic_model
```

```python
logistic_model.fit(X_train , y_train)
```

## 7. Model Testing

### Training Data

```python
y_pred = logistic_model.predict(X_train)
```

### Test Data
